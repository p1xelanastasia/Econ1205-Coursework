```python
# load libraries
import pandas as pd
import sqlite3
import os
```

```python
# load data
df_gdp = pd.read_csv("cross_sectional_gdp_2023.csv", dtype = str, keep_default_na = False)
df_unp = pd.read_csv("cross_sectional_unemployment_2023.csv", dtype = str, keep_default_na = False)
```

```python
# create database
conn = sqlite3.connect("wdi_cross_sectional_gdp_unemployment_2023.db")

# store data in database
df_gdp.to_sql("gdp", conn, if_exists = "replace", index = False)
df_unp.to_sql("unemployment", conn, if_exists = "replace", index = False)

# close connection
conn.close()
```
